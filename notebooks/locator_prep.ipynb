## Gambiae

```python
meta_gamb = meta[meta.taxon == 'gambiae']
```

```python
meta[meta.taxon == 'gambiae'].groupby('country').size()
```

```python
gamb_country = ['Burkina Faso', 'Cameroon', 'Central African Republic', 'Democratic Republic of the Congo', 'Gabon', 'Guinea', 'Mali', 'Mozambique', 'Tanzania', 'Uganda']
```

```python
ref_gamb_l_idxs = [random.sample(list(meta_gamb[meta_gamb.country == c].index), 50) for c in gamb_country]
for l in ref_gamb_l_idxs:
    l.sort()
ref_gamb_idxs = [s for l in ref_gamb_l_idxs for s in l]
```

```python
ag3.snp_calls('3L')
```

```python
ref_gamb_gt = ag3.snp_calls('2R')['call_genotype'][:,ref_gamb_idxs,:]
```

```python
ref_gamb_gt = ref_gamb_gt.chunk((ref_gamb_gt.chunks[0],tuple([50]*len(gamb_country)),ref_gamb_gt.chunks[2]))
```

```python
ref_gamb_gt.chunks[1]
```

```python
ref_gamb_gt
```

```python
meta_ref_gamb = meta_gamb.loc[ref_gamb_idxs]
long_l = []
lat_l = []
for i in range(0, 500, 100):
    long_l = long_l + ["NA"] * 3 + [str(l) for l in meta_gamb.loc[ref_gamb_idxs].longitude[i+3:i+100]]
    lat_l = lat_l + ["NA"] * 3 + [str(l) for l in meta_gamb.loc[ref_gamb_idxs].latitude[i+3:i+100]]
meta_ref_gamb['new_longitude'] = long_l
meta_ref_gamb['new_latitude'] = lat_l
gamb_metadata_exec = meta_ref_gamb[['new_longitude', 'new_latitude', 'sample_id']].rename(columns={'sample_id': 'sampleID', 'new_longitude': 'x', 'new_latitude': 'y'})
```
